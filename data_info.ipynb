#### `labeled_data`에서 제거된 컬럼 확인

```python
del_labeled = sorted(list(set(labeled.columns) - set(supervised_cn7.columns)))

for c in del_labeled:
    if c in ['_id', 'TimeStamp', 'PART_FACT_PLAN_DATE']:
        continue
    print(f"========== {c} ==========")
    print(labeled[c].value_counts())
    print("\n")
```

- `_id` : 제조 공정 id **-> 불필요**
- `TimeStamp` : 시간 **-> 불필요**
- `PART_FACT_PLAN_DATE` : 생산을 지시한 날짜 **-> 불필요**
- `PART_FACT_SERIAL` : 생산을 지시한 품목에 대한 코드 부여 **-> 불필요**
- `PART_NAME` : 제품의 이름 **-> CN7 제품만 선택**
- `EQUIP_CD` : 생산한 사출기 호기 **-> S14만 선택**
- `EQUIP_NAME` : 생산한 사출기 모델명 **-> 650톤-우진2호기만 선택**
- `Reason` : 불량 유형 **-> 불필요, 결측치 많음**
- `Switch_Over_Position` : 고압(사출압)에서 보압으로 진행될 때의 위치 **-> 대부분 값이 0**
- `Barrel_Temperature_7` : 계량 및 사출 시 수지가 일정하게 용융(녹임)을 유지하기 위해 온도가 일정해야 한다 **-> 모든 값이 0**
- `Mold_Temperature_1~2, 5~12` : 재료 주입구의 온도 **-> 모든 값이 0**

```python
labeled.groupby('PART_NAME')[['EQUIP_CD', 'EQUIP_NAME']].value_counts()
```

#### `labeled_data.csv` 전처리 후 `supervised_label_cn7.csv`와 비교

```python
# CN7 제품만 선택
labeled_processed = labeled[labeled['PART_NAME'].isin(["CN7 W/S SIDE MLD'G LH", "CN7 W/S SIDE MLD'G RH"])].copy()

# 불필요한 컬럼 제거
labeled_processed.drop(columns=del_labeled, inplace=True)

# PassOrFail 인코딩
labeled_processed['PassOrFail'] = labeled_processed['PassOrFail'].map({'Y': 0, 'N': 1})
print(labeled_processed['PassOrFail'].unique())
print(labeled_processed['PassOrFail'].value_counts())

supervised_cn7.equals(labeled_processed)
```

```python
print("Shape:", labeled_processed.shape == supervised_cn7.shape)
print("Columns:", labeled_processed.columns.equals(supervised_cn7.columns))
print("Index:", labeled_processed.index.equals(supervised_cn7.index))
print("Values:", labeled_processed.equals(supervised_cn7))
```

```python
# labeled_data는 숫자형 컬럼의 소수점이 9자리까지 있고, supervised_label_cn7은 컬럼마다 다름

for decimals in range(10, 0, -1):

    df1 = labeled_processed.copy()
    df2 = supervised_cn7.copy()

    num_cols = df1.select_dtypes(include="number").columns

    df1[num_cols] = df1[num_cols].round(decimals)
    df2[num_cols] = df2[num_cols].round(decimals)

    vc1 = df1.value_counts(dropna=False).sort_index()
    vc2 = df2.value_counts(dropna=False).sort_index()

    print(
        f"round({decimals}):",
        vc1.equals(vc2)
    )
```

```python
# 소수점 6자리까지 반올림해서 다시 비교
df1 = labeled_processed.copy()
df2 = supervised_cn7.copy()

num_cols = df1.select_dtypes(include="number").columns

df1[num_cols] = df1[num_cols].round(6)
df2[num_cols] = df2[num_cols].round(6)

vc1 = df1.value_counts(dropna=False).sort_index()
vc2 = df2.value_counts(dropna=False).sort_index()

print(vc1.equals(vc2))
```

#### `moldset_labeled.csv` 전처리 후 `moldset_labeled_cn7/rg3.csv`와 비교

```python
moldset_labeled['PassOrFail'].unique()
```

```python
# 제품 선택
moldset_processed_cn7 = moldset_labeled[moldset_labeled['PART_NAME'].isin(["CN7 W/S SIDE MLD'G LH", "CN7 W/S SIDE MLD'G RH"])].copy()
moldset_processed_rg3 = moldset_labeled[moldset_labeled['PART_NAME'].isin(["RG3 MOLD'G W/SHLD, LH", "RG3 MOLD'G W/SHLD, RH"])].copy()

# 불필요한 컬럼 제거
moldset_processed_cn7.drop(columns=del_moldset, inplace=True)
moldset_processed_rg3.drop(columns=del_moldset, inplace=True)

# PassOrFail 인코딩 불필요

print("moldset_processed_cn7.shape", moldset_processed_cn7.shape)
print("moldset_processed_rg3.shape", moldset_processed_rg3.shape)
```

> `moldset_labeled_cn7`은 (1211, 25)인데 `moldset_labeled_cn7.shape`는 (1425, 25)

```python
print("CN7 양품 수:", moldset_processed_cn7[moldset_processed_cn7['PassOrFail']==0].shape[0])
print("CN7 불량품 수:", moldset_processed_cn7[moldset_processed_cn7['PassOrFail']==1].shape[0])
print("="*20)
print("RG3 양품 수:", moldset_processed_rg3[moldset_processed_rg3['PassOrFail']==0].shape[0])
print("RG3 불량품 수:", moldset_processed_rg3[moldset_processed_rg3['PassOrFail']==1].shape[0])
```

```python
from sklearn.preprocessing import StandardScaler
for column in moldset_processed_rg3.columns:
    if column != 'PassOrFail':
        sc = StandardScaler()
        moldset_processed_rg3[column] = sc.fit_transform(moldset_processed_rg3[[column]])
```

```python
# moldset_labeled_rg3.csv만 비교해보기

for decimals in range(15, 0, -1):

    df1 = labeled_rg3.copy()
    df2 = moldset_processed_rg3.copy()

    num_cols = df1.select_dtypes(include="number").columns

    df1[num_cols] = df1[num_cols].round(decimals)
    df2[num_cols] = df2[num_cols].round(decimals)

    vc1 = df1.value_counts(dropna=False).sort_index()
    vc2 = df2.value_counts(dropna=False).sort_index()

    print(
        f"round({decimals}):",
        vc1.equals(vc2)
    )
```
